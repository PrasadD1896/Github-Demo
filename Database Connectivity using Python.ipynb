```python
import mysql.connector
```

```python
#Estabilish the connection

cnx = mysql.connector.connect(
    host = 'localhost',
    user = 'root',
    password = 'root',
    database = 'stored_procedure_demo'
)
```

```python
cursor = cnx.cursor() #Create cursor object
```

```python
query = 'select * from employees'
```

```python
cursor.execute(query)
```

```python
cnx.commit()
```

```python
for row in cursor.fetchall(): #fetchone , fetchall , fetchmany(size)
	print(row)
```

```python
query = 'insert into employees (id,first_name,last_name,salary) values (%s,%s,%s,%s)'
values = (4,'Rohit','Patil',85000)
```

```python
cursor.execute(query,values)
```

```python
cnx.commit()
```

```python
cursor.execute('select * from employees')
```
